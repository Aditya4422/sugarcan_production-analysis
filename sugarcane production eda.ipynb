## Data cleaning

```python
# Clean and convert numeric columns with proper comma and dot handling
df["Production(Tons)"] = df["Production(Tons)"].astype(str).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)
df["Production_per_Person(Kg)"] = df["Production_per_Person(Kg)"].astype(str).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)
df["Acreage(Hectare)"] = df["Acreage(Hectare)"].astype(str).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)
df["Yield(Kg/Hectare)"] = df["Yield(Kg/Hectare)"].astype(str).str.replace(".", "", regex=False).str.replace(",", ".", regex=False)

# Convert cleaned strings to numeric
df["Production(Tons)"] = pd.to_numeric(df["Production(Tons)"], errors="coerce")
df["Production_per_Person(Kg)"] = pd.to_numeric(df["Production_per_Person(Kg)"], errors="coerce")
df["Acreage(Hectare)"] = pd.to_numeric(df["Acreage(Hectare)"], errors="coerce")
df["Yield(Kg/Hectare)"] = pd.to_numeric(df["Yield(Kg/Hectare)"], errors="coerce")
```

```python
df.head()
```

```python
df.rename(columns={"Production (Tons)":"Production(Tons)"},inplace= True)
df.rename(columns={"Production per Person (Kg)":"Production_per_Person(Kg)"},inplace= True)
df.rename(columns={"Acreage (Hectare)":"Acreage(Hectare)"},inplace= True)
df.rename(columns={"Yield (Kg / Hectare)":"Yield(Kg/Hectare)"},inplace= True)
```

```python

df_pr = df["Continent"].value_counts()
plt.bar(df_pr.index, df_pr.values)
plt.xlabel("Continent")
plt.ylabel("Count")
plt.title("Sugarcane Production Entries by Continent")
plt.xticks(rotation=45)  # Optional: rotate x-axis labels
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
sns.histplot(df["Production(Tons)"], kde=True)
plt.title("Production (Tons)")

plt.subplot(2, 2, 2)
sns.histplot(df["Production_per_Person(Kg)"], kde=True)
plt.title("Production per Person (Kg)")

plt.subplot(2, 2, 3)
sns.histplot(df["Acreage(Hectare)"], kde=True)
plt.title("Acreage (Hectare)")

plt.subplot(2, 2, 4)
sns.histplot(df["Yield(Kg/Hectare)"], kde=True)
plt.title("Yield(Kg/Hectare)")

plt.tight_layout()
plt.show()
```
